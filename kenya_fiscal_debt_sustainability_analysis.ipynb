```python
# ============================================
# STEP 1: LOAD AND INSPECT THE CBK DATASETS
# ============================================

# Import pandas for data handling
import pandas as pd

# --------------------------------------------
# Load the three CSV files
# --------------------------------------------

public_debt = pd.read_csv("565446111_Public Debt.csv")
domestic_debt = pd.read_csv("1263030173_Domestic Debt by Instrument.csv")
revenue_expenditure = pd.read_csv("2118703754_Revenue and Expenditure.csv")

# --------------------------------------------
# Check the shape of each dataset
# Shape shows: number of rows and columns
# --------------------------------------------

print("Public Debt shape:", public_debt.shape)
print("Domestic Debt shape:", domestic_debt.shape)
print("Revenue and Expenditure shape:", revenue_expenditure.shape)

# --------------------------------------------
# Display the first few rows of each dataset
# This helps us understand how the data is structured
# --------------------------------------------

print("\nPUBLIC DEBT DATA")
display(public_debt.head(10))

print("\nDOMESTIC DEBT BY INSTRUMENT DATA")
display(domestic_debt.head(10))

print("\nREVENUE AND EXPENDITURE DATA")
display(revenue_expenditure.head(10))

# --------------------------------------------
# Check column names
# This helps us see if the data has clean headers
# --------------------------------------------

print("\nPublic Debt Columns:")
print(public_debt.columns)

print("\nDomestic Debt Columns:")
print(domestic_debt.columns)

print("\nRevenue and Expenditure Columns:")
print(revenue_expenditure.columns)

# --------------------------------------------
# Check data types
# This tells us whether numbers are stored correctly
# Most CBK CSV files usually load numbers as text
# because of commas like 503,554.00
# --------------------------------------------

print("\nPublic Debt Data Types:")
print(public_debt.dtypes)

print("\nDomestic Debt Data Types:")
print(domestic_debt.dtypes)

print("\nRevenue and Expenditure Data Types:")
print(revenue_expenditure.dtypes)
```

```python
# ============================================
# STEP 2: CLEAN PUBLIC DEBT DATASET
# ============================================

# Load the Public Debt dataset correctly
# skiprows=3 removes the title rows and starts from the real header row
public_debt_clean = pd.read_csv(
    "565446111_Public Debt.csv",
    skiprows=3
)

# --------------------------------------------
# Clean column names
# Remove extra spaces from column names
# --------------------------------------------
public_debt_clean.columns = public_debt_clean.columns.str.strip()

# Rename columns into clean analysis-friendly names
public_debt_clean = public_debt_clean.rename(columns={
    "Domestic Debt": "Domestic_Debt",
    "External Debt": "External_Debt",
    "Total": "Total_Public_Debt"
})

# --------------------------------------------
# Convert numeric columns
# Remove commas then convert to numbers
# --------------------------------------------
debt_columns = ["Domestic_Debt", "External_Debt", "Total_Public_Debt"]

for col in debt_columns:
    public_debt_clean[col] = (
        public_debt_clean[col]
        .astype(str)
        .str.replace(",", "", regex=False)
    )
    public_debt_clean[col] = pd.to_numeric(public_debt_clean[col], errors="coerce")

# --------------------------------------------
# Convert Year and Month into numeric values
# --------------------------------------------
public_debt_clean["Year"] = pd.to_numeric(public_debt_clean["Year"], errors="coerce")
public_debt_clean["Month"] = pd.to_numeric(public_debt_clean["Month"], errors="coerce")

# --------------------------------------------
# Create a proper Date column
# We use day=1 because the dataset only has year and month
# --------------------------------------------
public_debt_clean["Date"] = pd.to_datetime(
    dict(
        year=public_debt_clean["Year"],
        month=public_debt_clean["Month"],
        day=1
    ),
    errors="coerce"
)

# --------------------------------------------
# Remove empty or invalid rows
# --------------------------------------------
public_debt_clean = public_debt_clean.dropna(subset=["Date"])

# --------------------------------------------
# Sort the data by date
# --------------------------------------------
public_debt_clean = public_debt_clean.sort_values("Date").reset_index(drop=True)

# --------------------------------------------
# Preview final cleaned dataset
# --------------------------------------------
print("Cleaned Public Debt Dataset Shape:")
print(public_debt_clean.shape)

display(public_debt_clean.head())
display(public_debt_clean.tail())

print("\nData Types:")
print(public_debt_clean.dtypes)

print("\nMissing Values:")
print(public_debt_clean.isnull().sum())
```

```python
# ============================================
# STEP 3: CLEAN DOMESTIC DEBT BY INSTRUMENT DATASET
# ============================================

# --------------------------------------------
# Load the dataset
# This file has title rows at the top, so we use skiprows=2
# The real column names start after the first two rows
# --------------------------------------------

domestic_debt_clean = pd.read_csv(
    "1263030173_Domestic Debt by Instrument.csv",
    skiprows=2
)

# --------------------------------------------
# Preview the raw loaded dataset
# --------------------------------------------

print("Raw Domestic Debt Dataset Preview:")
display(domestic_debt_clean.head())

# --------------------------------------------
# Rename columns into clean names
# These names are easier to use in Python, SQL, and Power BI
# --------------------------------------------

domestic_debt_clean.columns = [
    "Period",
    "Treasury_Bills",
    "Treasury_Bonds",
    "Government_Stocks",
    "Overdraft_CBK",
    "Commercial_Bank_Advances",
    "Other_Domestic_Debt",
    "Total_Domestic_Debt"
]

# --------------------------------------------
# Remove empty rows
# Some rows may be blank because of formatting in the CBK CSV
# --------------------------------------------

domestic_debt_clean = domestic_debt_clean.dropna(subset=["Period"])

# --------------------------------------------
# Convert Period into a proper Date column
# Example:
# Sep-99 becomes 1999-09-01
# --------------------------------------------

domestic_debt_clean["Date"] = pd.to_datetime(
    domestic_debt_clean["Period"],
    format="%b-%y",
    errors="coerce"
)

# --------------------------------------------
# Fix years that may be misread by Python
# This is important for old dates like Sep-99
# --------------------------------------------

domestic_debt_clean["Date"] = domestic_debt_clean["Date"].apply(
    lambda x: x.replace(year=x.year - 100) if pd.notnull(x) and x.year > 2026 else x
)

# --------------------------------------------
# Create Year and Month columns from Date
# These will help when merging with the Public Debt dataset
# --------------------------------------------

domestic_debt_clean["Year"] = domestic_debt_clean["Date"].dt.year
domestic_debt_clean["Month"] = domestic_debt_clean["Date"].dt.month

# --------------------------------------------
# Convert all debt instrument columns into numeric values
# The original values have commas, so Python reads them as text
# --------------------------------------------

instrument_columns = [
    "Treasury_Bills",
    "Treasury_Bonds",
    "Government_Stocks",
    "Overdraft_CBK",
    "Commercial_Bank_Advances",
    "Other_Domestic_Debt",
    "Total_Domestic_Debt"
]

for col in instrument_columns:
    domestic_debt_clean[col] = (
        domestic_debt_clean[col]
        .astype(str)
        .str.replace(",", "", regex=False)
    )
    
    domestic_debt_clean[col] = pd.to_numeric(
        domestic_debt_clean[col],
        errors="coerce"
    )

# --------------------------------------------
# Remove rows where Date could not be created
# --------------------------------------------

domestic_debt_clean = domestic_debt_clean.dropna(subset=["Date"])

# --------------------------------------------
# Sort data by Date
# --------------------------------------------

domestic_debt_clean = domestic_debt_clean.sort_values("Date").reset_index(drop=True)

# --------------------------------------------
# Preview the final cleaned dataset
# --------------------------------------------

print("Cleaned Domestic Debt Dataset Shape:")
print(domestic_debt_clean.shape)

display(domestic_debt_clean.head())
display(domestic_debt_clean.tail())

# --------------------------------------------
# Check data types
# --------------------------------------------

print("\nData Types:")
print(domestic_debt_clean.dtypes)

# --------------------------------------------
# Check missing values
# --------------------------------------------

print("\nMissing Values:")
print(domestic_debt_clean.isnull().sum())
```

```python
# ============================================
# STEP 4: CLEAN REVENUE & EXPENDITURE DATASET
# ============================================

# --------------------------------------------
# Load dataset using the REAL header row
# Row 5 contains the actual usable column names
# --------------------------------------------

revenue_exp_clean = pd.read_csv(
    "2118703754_Revenue and Expenditure.csv",
    skiprows=5
)

# --------------------------------------------
# Preview dataset
# --------------------------------------------

display(revenue_exp_clean.head())

# --------------------------------------------
# Clean column names
# --------------------------------------------

revenue_exp_clean.columns = (
    revenue_exp_clean.columns
    .astype(str)
    .str.strip()
    .str.replace(" ", "_")
)

# --------------------------------------------
# Display actual columns
# --------------------------------------------

print(revenue_exp_clean.columns)

# --------------------------------------------
# Rename important columns
# --------------------------------------------

revenue_exp_clean = revenue_exp_clean.rename(columns={
    "Unnamed:_0": "Year",
    "Unnamed:_1": "Month",
    "TOTAL": "Total_Tax_Revenue",
    "REVENUE": "Total_Revenue",
    "TOTAL.1": "Total_Recurrent_Expenditure",
    "TOTAL_EXPENDITURE": "Total_Expenditure"
})

# --------------------------------------------
# Remove empty rows
# --------------------------------------------

revenue_exp_clean = revenue_exp_clean.dropna(subset=["Year"])

# --------------------------------------------
# Convert Year and Month into numeric
# --------------------------------------------

revenue_exp_clean["Year"] = pd.to_numeric(
    revenue_exp_clean["Year"],
    errors="coerce"
)

revenue_exp_clean["Month"] = pd.to_numeric(
    revenue_exp_clean["Month"],
    errors="coerce"
)

# --------------------------------------------
# Create proper Date column
# --------------------------------------------

revenue_exp_clean["Date"] = pd.to_datetime(
    dict(
        year=revenue_exp_clean["Year"],
        month=revenue_exp_clean["Month"],
        day=1
    ),
    errors="coerce"
)

# --------------------------------------------
# Convert all possible numeric columns
# --------------------------------------------

for col in revenue_exp_clean.columns:

    # Skip date columns
    if col == "Date":
        continue

    revenue_exp_clean[col] = (
        revenue_exp_clean[col]
        .astype(str)
        .str.replace(",", "", regex=False)
    )

    revenue_exp_clean[col] = pd.to_numeric(
        revenue_exp_clean[col],
        errors="coerce"
    )

# --------------------------------------------
# Remove rows with invalid dates
# --------------------------------------------

revenue_exp_clean = revenue_exp_clean.dropna(subset=["Date"])

# --------------------------------------------
# Sort dataset
# --------------------------------------------

revenue_exp_clean = revenue_exp_clean.sort_values("Date")

# --------------------------------------------
# Reset index
# --------------------------------------------

revenue_exp_clean = revenue_exp_clean.reset_index(drop=True)

# --------------------------------------------
# Final preview
# --------------------------------------------

print("Dataset Shape:")
print(revenue_exp_clean.shape)

display(revenue_exp_clean.head())

print("\nData Types:")
print(revenue_exp_clean.dtypes)

print("\nMissing Values:")
print(revenue_exp_clean.isnull().sum())
```

```python
# ============================================
# FIX FINAL COLUMN NAMES
# ============================================

revenue_exp_clean = revenue_exp_clean.rename(columns={

    "Import": "Import_Duty",
    "Excise": "Excise_Duty",
    "Income": "Income_Tax",

    "OTHER": "Other_Tax_Income",

    "TAX": "Total_Tax_Revenue",

    "Total_Revenue": "Non_Tax_Revenue",

    "REVENUE.1": "Total_Revenue_And_Grants",

    "Programme": "Programme_Grants",

    "Total": "Total_Grants",

    "EXPENDITURE": "Development_Expenditure",

    "Total_Recurrent_Expenditure": "Total_Expenditure"
})

# --------------------------------------------
# Preview corrected columns
# --------------------------------------------

print(revenue_exp_clean.columns)
```

```python
# ============================================
# STEP 5: MERGE ALL DATASETS
# ============================================

# --------------------------------------------
# First inspect merge keys
# --------------------------------------------

print("Public Debt Columns")
print(public_debt_clean.columns)

print("\nDomestic Debt Columns")
print(domestic_debt_clean.columns)

print("\nRevenue & Expenditure Columns")
print(revenue_exp_clean.columns)

# ============================================
# MERGE PUBLIC DEBT + DOMESTIC DEBT
# ============================================

master_fiscal_df = pd.merge(
    public_debt_clean,
    domestic_debt_clean,
    on=["Year", "Month", "Date"],
    how="left"
)

# --------------------------------------------
# Preview intermediate merge
# --------------------------------------------

print("\nAfter First Merge:")
print(master_fiscal_df.shape)

display(master_fiscal_df.head())

# ============================================
# MERGE REVENUE & EXPENDITURE DATA
# ============================================

master_fiscal_df = pd.merge(
    master_fiscal_df,
    revenue_exp_clean,
    on=["Year", "Month", "Date"],
    how="left"
)

# --------------------------------------------
# Final merged dataset preview
# --------------------------------------------

print("\nFinal Master Dataset Shape:")
print(master_fiscal_df.shape)

display(master_fiscal_df.head())

# ============================================
# CHECK DUPLICATE ROWS
# ============================================

duplicates = master_fiscal_df.duplicated().sum()

print("\nDuplicate Rows:")
print(duplicates)

# ============================================
# CHECK MISSING VALUES
# ============================================

print("\nMissing Values:")
print(master_fiscal_df.isnull().sum())

# ============================================
# DATASET INFO
# ============================================

print("\nMaster Dataset Info:")
print(master_fiscal_df.info())
```

```python
# ============================================
# STEP 6: CREATE FISCAL KPIs & ECONOMIC METRICS
# ============================================

# Create a clean copy of the merged dataset
kpi_df = master_fiscal_df.copy()

# ============================================
# REMOVE DUPLICATE COLUMNS
# CBK datasets create duplicate names after merges
# ============================================

kpi_df = kpi_df.loc[:, ~kpi_df.columns.duplicated()]

# ============================================
# CREATE KPIs
# ============================================

# --------------------------------------------
# 1. Fiscal Deficit
# Total Expenditure - Total Revenue & Grants
# --------------------------------------------

kpi_df["Fiscal_Deficit"] = (
    kpi_df["Total_Expenditure"]
    - kpi_df["Total_Revenue_And_Grants"]
)

# --------------------------------------------
# 2. Total Interest Payments
# Domestic Interest + Foreign Interest
# --------------------------------------------

kpi_df["Total_Interest_Payments"] = (
    kpi_df["Domestic_Interest"]
    + kpi_df["Foreign_Interest"]
)

# --------------------------------------------
# 3. Debt Growth Rate %
# --------------------------------------------

kpi_df["Debt_Growth_Rate_%"] = (
    kpi_df["Total_Public_Debt"]
    .pct_change()
    * 100
)

# --------------------------------------------
# 4. Revenue Growth Rate %
# --------------------------------------------

kpi_df["Revenue_Growth_Rate_%"] = (
    kpi_df["Total_Revenue_And_Grants"]
    .pct_change()
    * 100
)

# --------------------------------------------
# 5. Expenditure Growth Rate %
# --------------------------------------------

kpi_df["Expenditure_Growth_Rate_%"] = (
    kpi_df["Total_Expenditure"]
    .pct_change()
    * 100
)

# --------------------------------------------
# 6. Debt to Revenue Ratio
# --------------------------------------------

kpi_df["Debt_to_Revenue_Ratio"] = (
    kpi_df["Total_Public_Debt"]
    / kpi_df["Total_Revenue_And_Grants"]
)

# --------------------------------------------
# 7. Interest Burden Ratio
# --------------------------------------------

kpi_df["Interest_Burden_Ratio"] = (
    kpi_df["Total_Interest_Payments"]
    / kpi_df["Total_Revenue_And_Grants"]
)

# --------------------------------------------
# 8. Domestic Debt Ratio
# --------------------------------------------

kpi_df["Domestic_Debt_Ratio"] = (
    kpi_df["Domestic_Debt"]
    / kpi_df["Total_Public_Debt"]
)

# --------------------------------------------
# 9. External Debt Ratio
# --------------------------------------------

kpi_df["External_Debt_Ratio"] = (
    kpi_df["External_Debt"]
    / kpi_df["Total_Public_Debt"]
)

# --------------------------------------------
# 10. Treasury Bills Share
# --------------------------------------------

kpi_df["Treasury_Bills_Share"] = (
    kpi_df["Treasury_Bills"]
    / kpi_df["Total_Domestic_Debt"]
)

# --------------------------------------------
# 11. Treasury Bonds Share
# --------------------------------------------

kpi_df["Treasury_Bonds_Share"] = (
    kpi_df["Treasury_Bonds"]
    / kpi_df["Total_Domestic_Debt"]
)

# ============================================
# ROUND NUMERIC COLUMNS
# ============================================

for col in kpi_df.select_dtypes(include=["number"]).columns:
    kpi_df[col] = kpi_df[col].round(2)

# ============================================
# KPI COLUMNS TO DISPLAY
# ============================================

kpi_columns = [
    "Date",
    "Fiscal_Deficit",
    "Total_Interest_Payments",
    "Debt_Growth_Rate_%",
    "Revenue_Growth_Rate_%",
    "Expenditure_Growth_Rate_%",
    "Debt_to_Revenue_Ratio",
    "Interest_Burden_Ratio",
    "Domestic_Debt_Ratio",
    "External_Debt_Ratio",
    "Treasury_Bills_Share",
    "Treasury_Bonds_Share"
]

# ============================================
# OUTPUT
# ============================================

print("KPI Dataset Shape:")
print(kpi_df.shape)

print("\nKPI Dataset Preview:")
display(kpi_df[kpi_columns].head(10))

# ============================================
# CHECK MISSING VALUES IN KPI COLUMNS
# ============================================

print("\nMissing Values in KPI Columns:")
print(kpi_df[kpi_columns].isnull().sum())

# ============================================
# SUMMARY STATISTICS
# ============================================

print("\nKPI Summary Statistics:")
display(kpi_df[kpi_columns].describe())
```
